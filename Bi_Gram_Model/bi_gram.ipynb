```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
words=open(r'C:\Users\arput\OneDrive\Documents\Learning_ML\project\ML_Simple_project\Bi_Gram_Model\names.txt').read()
```

```python
stoi={}
itos={}
word=words.splitlines()
w=words.replace('\n','.')
w=list(set(w))
w.sort()
print(len(w))
```

```python
len(word)*0.9
```

```python
for i in range(0,len(w)):
    stoi[w[i]]=i
    itos[i]=w[i]

N = np.zeros((27, 27), dtype=int)
for name in word[:28830]:
    s = "." + name + "."
    for a, b in zip(s, s[1:]):
        N[stoi[a], stoi[b]] += 1
```

```python
N=N+1
```

```python
stoi
```

```python
prob=N/N.sum(axis=1,keepdims=True)
```

```python
total=0
count=0
for name in word:
    s = "." + name + "."
    for a, b in zip(s, s[1:]):
        total+=-np.log(prob[stoi[a],stoi[b]])
        count+=1
print(total/count)
```
