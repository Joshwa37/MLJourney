```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
input=np.array([[0,1],[1,0],[0,0],[1,1]])
value=np.array([[0],[1],[1],[0]])
```

```python
weight=np.zeros([8,2])
```

```python
weight=weight.T
bias=np.ones([1,8])
weight2=np.zeros([8,1])
bias2=np.array([1])
```

```python
for i in range(20000):
    h1=input@weight+bias
    act=np.tanh(h1)
    h2=act@weight2+bias2
    loss=np.sum((h2-value)**2)/4
    print(loss)
    def backwardpass():
        dh2=0
        dact=0
        dweight2=0
        dbias2=0
        dtanh=0
        dh1=0
        dinput=0
        dweight=0
        dbias=0
        dh2+=(h2-value)/2
        dact+=dh2@weight2.T
        dweight2+=act.T@dh2
        dbias2+=np.sum(dh2,axis=0)
        dtanh+=4/(np.exp(h1)+np.exp(-h1))**2
        dh1+=dact*dtanh
        dinput+=dh1@weight.T
        dweight+=input.T@dh1
        dbias+=np.sum(dh1,axis=0)

        return dbias,dweight,dbias2,dweight2
    dbias,dweight,dbias2,dweight2=backwardpass()
    lr=0.01
    weight=weight-(lr*dweight)
    bias=bias-(lr*dbias)
    weight2=weight2-(lr*dweight2)
    bias2=bias2-(lr*dbias2)

    if(loss<0):
        break
    realloss=loss
```
